```python
from pyspark.sql.functions import col, count, when

# 1. THE CLEANING FUNCTION
def clean_nulls(df, name):
    print(f"Processing {name}...")
    
    # Check nulls before
    null_counts_before = df.select([count(when(col(c).isNull(), c)).alias(c) for c in df.columns]).collect()[0].asDict()
    total_nulls_before = sum(null_counts_before.values())
    
    if total_nulls_before > 0:
        print(f"  - Found {total_nulls_before} total null values.")
        
        # Fill numeric nulls with 0 and string nulls with "0"
        # Note: Dates are usually left as null or handled specifically, 
        # but we'll target strings and numbers here.
        df_cleaned = df.na.fill(0)      # Fills Integer/Double
        df_cleaned = df_cleaned.na.fill("0") # Fills Strings
        
        # Check nulls after
        null_counts_after = df_cleaned.select([count(when(col(c).isNull(), c)).alias(c) for c in df_cleaned.columns]).collect()[0].asDict()
        total_nulls_after = sum(null_counts_after.values())
        
        print(f"  - Fixed! Nulls remaining: {total_nulls_after} (Dates stay null to prevent errors)")
        return df_cleaned
    else:
        print("  - No nulls found. Table is already clean.")
        return df

# 2. APPLY TO ALL TABLES
cleaned_dataframes = {}
for name, df in all_dataframes.items():
    cleaned_dataframes[name] = clean_nulls(df, name)

# 3. VERIFY THE BIG ONE (Sales Pipeline)
print("\n📊 Verification: Sales Pipeline (close_value)")
cleaned_dataframes['sales_pipeline'].select("opportunity_id", "deal_stage", "close_value").show(5)
```

```python
from pyspark.sql import functions as F
from pyspark.sql.types import DateType

# 1. THE "TOTAL CLEAN" FUNCTION
def total_null_removal(name, df):
    print(f"🧹 Processing {name}...")
    
    # Identify Date columns
    date_cols = [f.name for f in df.schema.fields if isinstance(f.dataType, DateType)]
    
    # Step A: Convert Date columns to String so they can accept "0"
    for col_name in date_cols:
        df = df.withColumn(col_name, F.col(col_name).cast("string"))
    
    # Step B: Fill numeric nulls with 0
    df = df.fillna(0)
    
    # Step C: Fill all string nulls (including the new Date strings) with "0"
    df = df.fillna("0")
    
    # Step D: Final Audit
    null_count = df.select([F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in df.columns])
    total_remaining = sum(null_count.collect()[0].asDict().values())
    
    print(f"✅ {name} complete. Total Nulls: {total_remaining}")
    return df

# 2. APPLY TO YOUR DATA
cleaned_spark_dfs = {}
for name, df in all_dataframes.items():
    cleaned_spark_dfs[name] = total_null_removal(name, df)

print("\n✨ ALL TABLES ARE NOW 100% NULL-FREE!")
```

```python
import csv
from pyspark.sql import functions as F

# 1. THE EXPORT FUNCTION (Bypasses all environment errors)
def save_spark_to_csv(df, filename):
    print(f"📦 Exporting {filename}...")
    
    # Get column names from the Spark DataFrame
    column_names = df.columns
    
    # Collect data to the driver (fine for CRM sized data)
    data_rows = df.collect()
    
    # Write using standard Python CSV module
    with open(filename, 'w', newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        writer.writerow(column_names) # Write Header
        for row in data_rows:
            writer.writerow(row) # Write Data Rows
            
    print(f"✅ Success: {filename} created.")

# 2. PERFORM CLEANING
cleaned_dfs = {}
for name, df in all_dataframes.items():
    # Fill nulls: Numeric -> 0, Strings/Dates -> "0"
    df_clean = df.fillna(0).fillna("0")
    cleaned_dfs[name] = df_clean

# 3. PERFORM CALCULATIONS (Join)
master_df = cleaned_dfs['sales_pipeline'] \
    .join(cleaned_dfs['sales_teams'], "sales_agent", "left") \
    .join(cleaned_dfs['products'], "product", "left") \
    .join(cleaned_dfs['accounts'], "account", "left") \
    .fillna("0") # Final safety check

# 4. EXPORT ALL FILES
# A. Export cleaned original datasets
for name, df in cleaned_dfs.items():
    save_spark_to_csv(df, f"cleaned_{name}.csv")

# B. Export the Master Join (The full CRM snapshot)
save_spark_to_csv(master_df, "analytics_master_joined.csv")

# C. Export Manager Revenue Table
manager_report = master_df.groupBy("manager").agg(
    F.sum("close_value").alias("total_revenue"),
    F.count("opportunity_id").alias("deal_count")
).orderBy(F.desc("total_revenue"))

save_spark_to_csv(manager_report, "analytics_manager_revenue.csv")

print("\n🚀 ALL CALCULATIONS DONE AND FILES EXPORTED!")
```
